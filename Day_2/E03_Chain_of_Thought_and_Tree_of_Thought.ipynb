<div class="topic-header">
<strong>Progress: Exercise 3 of 15 · Day 1</strong> &nbsp;|&nbsp; Facilitator-run — you watch,
predict, and judge the outputs. No coding required.
<br><br>
<strong>Why this matters to you as an engineering manager:</strong> Most of the decisions that
cross your desk are not lookups — they are <em>reasoning problems</em>: how many pods do we
need for peak-season traffic, why did checkout latency spike after last night's deploy, do we
refactor the order-management monolith or rewrite it, which region gets the constrained
inventory. In E02 we saw that a model answering such questions <em>leaps straight to an
answer</em> — confident, fluent, and often wrong. This exercise gives you the two techniques
that fix that: <strong>Chain-of-Thought</strong> (force the model to reason in steps) and
<strong>Tree-of-Thought</strong> (force it to reason from multiple independent perspectives
and reconcile them). Your job today is not to write prompts — it is to learn to
<em>recognize which of your team's problems need which technique</em>, so you can set the
standard for how your teams use these tools.
<br><br>
<strong>Duration:</strong> ~45 minutes &nbsp;|&nbsp;
<strong>Builds on:</strong> E02 — Zero-Shot &amp; Few-Shot Prompting
</div>

## 🔓 Picking up E02's open gap

<div class="warning-box">
<strong>Where E02 left us:</strong> Zero-shot and few-shot prompting worked beautifully for
classification and formatting — tasks where the model already "knows" the answer and just
needs the right output shape. But we ended E02 on an unresolved problem: on
<strong>multi-step reasoning</strong> — capacity math, root-cause analysis, trade-off
decisions — the model <em>leaps to an answer</em> without working through the steps, and the
leap is frequently wrong. Examples alone cannot fix that, because the failure is not about
output format — it is about the model never doing the intermediate thinking at all.
</div>

<div class="fix-box">
<strong>What E03 adds:</strong> Reasoning problems need the model to <em>think in steps</em>.
Chain-of-Thought makes each intermediate step explicit text the model can build on;
Tree-of-Thought runs several independent reasoning paths and reconciles them. Neither gives
the model new knowledge — they give it a structured way to <em>use</em> the knowledge it has.
</div>

## Learning Objectives

By the end of this exercise you will be able to:

1. **Explain why** LLMs produce wrong answers on multi-step problems when prompted directly
2. **Recognize Chain-of-Thought prompting** (zero-shot and few-shot) applied to capacity, cost, and scheduling problems
3. **Distinguish** when zero-shot CoT is enough vs when a worked example (few-shot CoT) is needed
4. **See CoT turn a generic incident checklist into a structured root-cause procedure** (checkout latency spike)
5. **Apply multi-hop reasoning** to questions that chain multiple facts (service-portfolio queries)
6. **Use Tree-of-Thought** for strategic engineering decisions (refactor vs rewrite) and cross-checked calculations (inventory/dock allocation)
7. **Choose the right technique** for a given problem type using a decision framework

<div class="concept-box">
<strong>🧠 Mental model for this exercise:</strong> Think of the model like a strong engineer
in an interview. Asked to answer <em>instantly</em>, they pattern-match and blurt something
plausible. Given a whiteboard — "walk me through it" — they lay out steps, catch their own
mistakes, and land on the right answer. CoT is the whiteboard. ToT is a design-review panel:
three engineers work the same problem independently, then a lead reconciles their views.
</div>

```python
!pip install -q openai tiktoken

import openai

API_KEY = 'PASTE_THE_KEY_SHARED_IN_SESSION_HERE'
MODEL   = 'gpt-3.5-turbo-0125' # gpt-5.4-nano

client = openai.OpenAI(api_key=API_KEY)

def ask(prompt, system='You are a helpful assistant.', temperature=0.3, max_tokens=1024):
    '''Send a prompt and return the model response text.'''
    r = client.chat.completions.create(
        model=MODEL,
        messages=[
            {'role': 'system', 'content': system},
            {'role': 'user',   'content': prompt}
        ],
        temperature=temperature,
        max_completion_tokens=max_tokens
    )
    return r.choices[0].message.content

def show(label, text):
    '''Pretty-print a labelled result.'''
    print()
    print('=' * 60)
    print(f'  {label}')
    print('=' * 60)
    print(text)
    print()

print('Setup complete!')
print(f'Model: {MODEL}')
```

## Part 1: The Problem — LLMs Rush to Wrong Answers

Let us start with a capacity-planning problem of the kind that lands on an engineering
manager's desk before every peak season. It contains a **decoy** (a capacity number that
looks relevant but is not), a **withdrawn requirement** the problem itself corrects, and a
**unit trap** — the forecast is per **hour** while pod capacity is per **minute**. Humans
sometimes fall for these; LLMs are even more susceptible because they generate answers
token-by-token without pausing to plan.

<div class="try-it">
<strong>The Problem:</strong><br>
Checkout pods are rated 400 orders/min but reserve 10% for health checks (360 usable). The
older search service's pods handle 250 orders/min. Last peak the site did 1,500,000
orders/hour; this year's forecast is 1,800,000 orders/<strong>hour</strong>, and the
requirement is 25% headroom above forecast. How many checkout pods are needed?<br><br>
<strong>The traps:</strong> The search-pod capacity (250) and last year's volume are
irrelevant. The "difference between capacities" headroom rule is withdrawn mid-problem.
And the model must convert hours to minutes (or pods to per-hour) consistently.<br><br>
<strong>🔮 Predict before we run:</strong> Will the model ignore the decoys AND get the
unit conversion right?
</div>

Let us see what happens when we just ask directly, with no reasoning instructions:

```python
# The tricky problem: decoy capacity, withdrawn rule, and an hours-vs-minutes unit trap
prompt = '''A large retailer's checkout service runs on pods rated at 400 orders per minute,
but each pod reserves 10% of its rated capacity for health checks and retries, so only the
remainder is usable for customer orders. The older search service's pods each handle 250
orders per minute, and last peak season the site processed 1,500,000 orders per hour. The
team compares all these figures during planning. This year's peak-season forecast is
1,800,000 orders per hour. The platform team initially asked for headroom equal to the
difference between the two pod capacities. Wait - actually the requirement is simply 25%
headroom above this year's forecast. How many checkout pods are needed for peak season?'''

result = ask(prompt)
show('Direct Answer (no reasoning instructions)', result)
```

<div class="where-box">
<strong>WHERE it breaks:</strong> The model sees numbers and jumps to an answer without
checking which numbers matter or whether the units line up. In the run above it tangled the
per-hour forecast with per-minute pod capacity and produced a wildly wrong pod count. On
other reruns it mixes the two pod capacities, applies headroom to the wrong base, or keeps
the "difference between capacities" idea the problem itself withdrew. <em>Live-demo note:</em>
this failure is frequent but not guaranteed — at temperature 0.3 the direct prompt gets
lucky on some reruns. If it does, run the cell again; the fragility is itself the point.
The correct answer is <strong>105 pods</strong> (worked out below).
</div>

<div class="why-box">
<strong>WHY this happens:</strong> LLMs generate text token by token. Each new token is
predicted from the tokens before it. Without explicit reasoning steps, the model has no
"scratchpad" — it must hold the entire solution internally and produce the answer in one
shot. This is like asking one of your engineers to size a fleet entirely in their head, out
loud, without a whiteboard. The more numbers, steps, and unit conversions involved, the more
likely the model loses track, gets distracted by decoys, or skips a step.
</div>

## Part 2: Chain-of-Thought Prompting — "Think Step by Step"

<div class="concept-box">
<strong>What is Chain-of-Thought (CoT)?</strong><br><br>
Chain-of-Thought prompting instructs the model to show its reasoning step by step before
giving a final answer. The key insight: each generated reasoning step becomes
<strong>context</strong> for the next step. The model can "see" its own earlier reasoning,
which prevents it from losing track of intermediate results.<br><br>
<strong>Why it works:</strong> When the model writes "Step 1: We are sizing checkout, so the
relevant capacity is 400 orders/min per pod" — that text is now in its context window. When
it generates Step 2, it can reference that result. Without CoT, the model has to hold all of
this internally, which is unreliable for multi-step problems.<br><br>
<strong>The simplest version:</strong> Just add <code>"Let us think step by step."</code> to
the prompt. This single phrase dramatically improves accuracy on math, logic, and reasoning
tasks.
</div>

<div class="fix-box">
<strong>The Fix:</strong> Same pod-sizing problem, but now we add one instruction: list the
relevant facts, discard corrected/irrelevant ones, and think step by step. Watch the model
ignore the decoy, show its work, and land on the right answer.
</div>

```python
# Same problem, but with CoT instruction
prompt_cot = '''A large retailer's checkout service runs on pods rated at 400 orders per minute,
but each pod reserves 10% of its rated capacity for health checks and retries, so only the
remainder is usable for customer orders. The older search service's pods each handle 250
orders per minute, and last peak season the site processed 1,500,000 orders per hour. The
team compares all these figures during planning. This year's peak-season forecast is
1,800,000 orders per hour. The platform team initially asked for headroom equal to the
difference between the two pod capacities. Wait - actually the requirement is simply 25%
headroom above this year's forecast. How many checkout pods are needed for peak season?

Before solving , first list which facts in the problem are relevant and which are irrelvant to the final answer. Note that any earlier statement that
is later corrected should be discarded,  think step by step. Solve and verify at each step based on the problem statement'''

result_cot = ask(prompt_cot)
show('With Chain-of-Thought', result_cot)
print('Expected: 1,800,000 orders/hour = 30,000/min; x 1.25 headroom = 37,500/min target')
print('Usable pod capacity = 400 x 0.90 = 360 orders/min')
print('37,500 / 360 = 104.17  =>  round up to 105 checkout pods')
print('Irrelevant / discarded: search pod capacity (250), last year 1,500,000/hr,')
print('and the withdrawn "difference between capacities" rule')
```

<div class="try-it">
<strong>Let us test CoT on three more problems</strong> — each with increasing complexity and
each containing some form of irrelevant or distracting information. For each problem we run
it twice: once without CoT, once with CoT. <strong>🔮 Predict before each run:</strong> which
fact is the decoy, and where will the direct answer go wrong?
</div>

### Problem 2: Percentage with Unnecessary Information

```python
# Problem 2: Percentage with a per-day unit trap and a percentage-of-which-base trap
problem_2 = '''A large retailer's engineering org receives 900 monitoring alerts per business day
(the org runs a 5-day on-call week). 60% of alerts relate to the storefront platform and
40% to fulfillment. The storefront platform has 15 microservices and was re-platformed in
2023. A new auto-remediation system resolves 35% of storefront alerts automatically.
How many alerts per week still need a human across the whole org, and what percentage of
the org's total weekly alerts is that?'''

# Without CoT
result_direct = ask(problem_2)
show('Problem 2 — WITHOUT CoT', result_direct)

# With CoT
result_cot = ask(problem_2 + '''\n\nBefore solving, first list which facts are relevant and which are irrelevant to the final
answer, and note any unit conversions needed (per-day vs per-week). Then think step by
step, and at the end re-read the question and confirm your final numbers answer exactly
what was asked, including which total any percentage is measured against.''')
show('Problem 2 — WITH CoT', result_cot)

print('Expected: 900/day x 5 days = 4,500 alerts/week')
print('Storefront: 4,500 x 0.60 = 2,700; auto-resolved: 2,700 x 0.35 = 945')
print('Still need a human: 4,500 - 945 = 3,555 alerts/week')
print('Percentage: 3,555 / 4,500 = 79% of all weekly alerts')
print('Irrelevant info: 15 microservices, re-platformed in 2023')
```

### Problem 3: Time-Zone Calculation (Incident Bridge)

```python
# Problem 3: Time-zone calculation (US Central home office to Bengaluru tech center)
problem_3 = '''A large retailer's home-office SRE team (US Central, UTC-5:00 during daylight
time) needs to schedule an incident-review bridge with the Bengaluru tech center
(IST, UTC+5:30). The US team has 12 members and the Bengaluru team has 8 members.
The bridge should start at 10:00 AM US Central time. What time is that in Bengaluru?
If the bridge lasts 45 minutes and there is a 15-minute break after it, what time does
the next meeting slot start in Bengaluru?'''

# Without CoT
result_direct = ask(problem_3)
show('Problem 3 — WITHOUT CoT', result_direct)

# With CoT
result_cot = ask(problem_3 + '''\n\nBefore solving , first list which facts in the problem are relevant and which are irrelvant to the final answer. Note that any earlier statement that
is later corrected should be discarded,  think step by step. Solve and verify at each step based on the problem statement''')
show('Problem 3 — WITH CoT', result_cot)

print('Expected: IST is 10.5 hours ahead of US Central (UTC-5:00).')
print('10:00 AM Central = 8:30 PM IST.')
print('Bridge ends 8:30 PM + 45 min = 9:15 PM IST.')
print('Break 15 min => next slot starts at 9:30 PM IST.')
print('Irrelevant info: team sizes (12 and 8 members)')
```

### Problem 4: Multi-Step Budget Calculation

```python
# Problem 4: Cloud budget chain with a monthly-vs-quarterly unit trap and an
# order-of-operations trap (compute % applies AFTER the shared charges are deducted)
problem_4 = '''A large retailer's e-commerce platform group has a quarterly cloud budget of
$2,400,000. The group also pays a shared-infrastructure charge billed at $50,000 per month.
Compute costs consume 65% of the budget that is left after the quarter's shared-
infrastructure charges are paid. Of what then remains, 25% is held back as a peak-season
contingency reserve. The rest is split equally across 3 domain teams. Team Checkout
commits 40% of its share to tech-debt reduction. How much does Team Checkout have left
for new feature infrastructure?'''

# Without CoT
result_direct = ask(problem_4)
show('Problem 4 — WITHOUT CoT', result_direct)

# With CoT (note: the CoT instruction spells out order-of-operations and unit checks)
result_cot = ask(problem_4 + '''\n\nSolve carefully step by step. First write out the exact order of operations the problem
implies (what is deducted first, which base each percentage applies to, and any monthly vs
quarterly unit conversions). Compute each quantity in that order, verifying each step
against the problem statement. Finally, re-read the question and confirm your final number
answers exactly what was asked - the amount LEFT for new feature infrastructure, not the
amount deducted.''')
show('Problem 4 — WITH CoT', result_cot)

print('Expected calculation:')
print('Budget                   = $2,400,000')
print('Shared infra (3 months)  = $150,000  =>  net $2,250,000')
print('Compute (65% of net)     = $1,462,500  =>  remaining $787,500')
print('Contingency (25%)        = $196,875  =>  distributable $590,625')
print('Each team                = $196,875')
print('Team Checkout features   = 60% of 196,875 = $118,125')
```

<div class="takeaway">
<strong>Pattern:</strong> The direct prompt gets distracted by irrelevant information, trips
on unit conversions (per-hour vs per-minute, monthly vs quarterly), or answers a different
question than the one asked. CoT forces the model to lay out each step explicitly, making
errors visible and correctable — exactly the property you want before an answer feeds a
staffing, budget, or capacity decision.
</div>

<div class="warning-box">
<strong>Honest calibration for your teams:</strong> On these problems the direct prompt does
not fail <em>every</em> time — modern models get lucky on some reruns, and CoT is not
infallible either (a single CoT chain can still misread a base or drop a charge). What CoT
reliably buys you is (1) a much higher hit rate and (2) an <em>auditable</em> chain: when it
is wrong, you can see exactly which step broke. When a single chain is not reliable enough,
run several independent chains and compare — that is exactly what Part 7 (multiple solvers)
does. Beware also of <em>badly worded</em> CoT instructions: telling the model to "discard
irrelevant facts" on a problem with no decoys can make it discard a real cost line.
</div>

## Part 3: Zero-Shot CoT vs Few-Shot CoT

<div class="concept-box">
<strong>Two flavours of Chain-of-Thought:</strong><br><br>
<strong>Zero-Shot CoT</strong> — You simply append an instruction like "Let us think step by
step". The model figures out <em>how</em> to reason on its own. This is what we used in
Part 2. It works well for straightforward reasoning tasks.<br><br>
<strong>Few-Shot CoT</strong> — You provide one or more <em>worked examples</em> that
demonstrate the exact reasoning pattern you want. The model follows that pattern for the new
problem. This is more powerful for domain-specific reasoning where the model might not know
the right "shape" of the solution.<br><br>
<strong>When does Zero-Shot CoT fail?</strong> On problems requiring a specific
problem-solving strategy — like two processes converging on a shared workload — the model may
start reasoning but follow the wrong approach. Few-Shot CoT fixes this by showing the correct
strategy via example.
</div>

Let us test with a **converging-jobs problem** — two batch jobs working toward each other
through the same dataset, the kind of estimate a manager needs for a migration cutover
window. Zero-shot CoT sometimes picks the wrong approach here:

```python
# A converging-jobs problem (relative-rate reasoning)
migration_problem = '''A reconciliation job starts at 7:15 AM processing an inventory ledger
of 150 GB from the head at 90 GB/hour. A validation job starts at 7:45 AM from the tail of
the same ledger at 110 GB/hour, working towards the first job. At what time do the two jobs
meet, and how many GB from the head of the ledger?'''

# Zero-shot CoT
result_zero = ask(migration_problem + '\n\nLet us think step by step.')
show('Zero-Shot CoT (just "think step by step")', result_zero)
```

```python
# Few-shot CoT: provide a worked example first
few_shot_prompt = '''Here is an example of how to solve converging-jobs problems:

EXAMPLE PROBLEM: A cleanup job starts on a 120 GB log archive at 8:00 AM processing from the
head at 60 GB/hour. A compression job starts at 8:30 AM from the tail at 80 GB/hour, working
towards the cleanup job. When do they meet?

EXAMPLE SOLUTION:
Step A: Find head start. By 8:30 AM, the cleanup job has processed 0.5h x 60 = 30 GB.
Step B: Remaining gap at 8:30 AM = 120 - 30 = 90 GB.
Step C: From 8:30 AM both jobs consume the gap. Combined closing rate = 60 + 80 = 140 GB/hour.
Step D: Time to close the 90 GB gap = 90 / 140 = 0.6429 hours = 38.57 minutes.
Step E: They meet at 8:30 AM + 38.57 min = approximately 9:09 AM.
Step F: Position from the head = 30 + (60 x 0.6429) = 30 + 38.57 = 68.57 GB.

---

NOW SOLVE THIS PROBLEM using the same step-by-step approach:

A reconciliation job starts at 7:15 AM processing an inventory ledger of 150 GB from the
head at 90 GB/hour. A validation job starts at 7:45 AM from the tail of the same ledger at
110 GB/hour, working towards the first job. At what time do the two jobs meet, and how many
GB from the head of the ledger?'''

result_few = ask(few_shot_prompt)
show('Few-Shot CoT (with worked example)', result_few)

print('Expected:')
print('By 7:45 AM, reconciliation processed 0.5h x 90 = 45 GB')
print('Remaining gap = 150 - 45 = 105 GB')
print('Combined rate = 90 + 110 = 200 GB/hour')
print('Time to meet = 105 / 200 = 0.525h = 31.5 min')
print('They meet at 7:45 + 31.5 min = 8:16:30 AM')
print('Position from head = 45 + (90 x 0.525) = 45 + 47.25 = 92.25 GB')
```

  <td><strong>Prompt effort</strong></td>
  <td>Minimal — add one sentence</td>
  <td>Higher — must craft worked examples</td>
</tr>
<tr>
  <td><strong>When it works</strong></td>
  <td>Standard math, logic, everyday reasoning</td>
  <td>Domain-specific reasoning, unusual problem types</td>
</tr>
<tr>
  <td><strong>When it fails</strong></td>
  <td>Model may pick the wrong strategy for unfamiliar problem types</td>
  <td>Rarely fails if examples are well-chosen</td>
</tr>
<tr>
  <td><strong>Token cost</strong></td>
  <td>Low (short prompt)</td>
  <td>Higher (prompt includes full worked example)</td>
</tr>
<tr>
  <td><strong>Best for</strong></td>
  <td>Quick improvement on arithmetic and logic</td>
  <td>Consistent format + complex domain reasoning</td>
</tr>
</table>

<div class="takeaway">
<strong>Rule of thumb:</strong> Start with zero-shot CoT. If the model consistently gets the
<em>approach</em> wrong (not just the arithmetic), switch to few-shot CoT with a worked
example that demonstrates the correct strategy — the same way you would hand a new engineer a
worked runbook rather than a one-line hint.
</div>

## Part 4: CoT for Incident Root-Cause Analysis

Chain-of-Thought is not just for math. One of its most powerful real-world applications is
**diagnostics** — where the difference between a generic list of possible causes and a
structured diagnostic procedure is the difference between a 3-hour war room and a 20-minute
targeted fix.

<div class="concept-box">
<strong>The Scenario:</strong> An incident ticket lands with your on-call team at a large
retailer:<br><br>
<em>"p99 checkout latency jumped from 280 ms to 2.1 s at 14:05. Error rate is flat. The spike
began about 10 minutes after a deploy of the cart-service. Database CPU is normal. Only
carts with more than 20 items appear affected."</em><br><br>
Without CoT, the model gives a "shotgun" list of every possible latency cause. With CoT, it
follows a logical diagnostic tree — checking the most likely cause first, explaining why, and
deciding the next step based on each result.
</div>

<div class="try-it">
<strong>🔮 Predict before we run:</strong> the ticket contains two strong clues (the recent
deploy, the &gt;20-item carts). Will the generic answer use them? Will the CoT answer?
</div>

```python
ticket = '''p99 checkout latency jumped from 280 ms to 2.1 s at 14:05. Error rate is
flat. The spike began about 10 minutes after a deploy of the cart-service. Database CPU
is normal. Only carts with more than 20 items appear affected. What should I check?'''

result_generic = ask(ticket)
show('Without CoT — Generic Troubleshooting', result_generic)
```

```python
# Structured diagnostic system prompt
diagnostic_system = '''You are a senior site reliability engineer at a large retailer with
15 years of experience running high-scale e-commerce systems. When troubleshooting, follow
this diagnostic methodology for EACH step:

(1) STATE what you are checking and exactly where (which dashboard, log, trace, or metric)
(2) EXPLAIN why you are checking this (what underlying system behavior would produce it)
(3) STATE what each possible result would tell you
(4) DECIDE the next step based on the most likely result

Work from the most likely cause to the least likely. Do not list all possibilities at once —
follow one diagnostic path at a time, like a real incident investigation. Anchor on the
clues in the ticket (recent changes, what still works, which segment is affected).'''

ticket = '''p99 checkout latency jumped from 280 ms to 2.1 s at 14:05. Error rate is flat.
The spike began about 10 minutes after a deploy of the cart-service. Database CPU is
normal. Only carts with more than 20 items appear affected.'''

result_diagnostic = ask(ticket, system=diagnostic_system)
show('With CoT System Prompt — Structured Diagnostics', result_diagnostic)
```

<div class="takeaway">
<strong>Key Difference:</strong><br>
<strong>Without CoT</strong> &rarr; "Here are 8 possible causes" (a <em>list of possibilities</em>)<br>
<strong>With CoT</strong> &rarr; "First check X because... if you see Y then... next check Z because..."
(a <em>diagnostic procedure</em>)<br><br>
CoT transforms the model from "a search engine that returns all results" into "a senior
engineer who walks the team through a logical investigation." For a manager, this is the
difference between an AI tool that adds noise to a war room and one that shortens it.
</div>

## Part 5: Multi-Hop Reasoning

<div class="concept-box">
<strong>What is Multi-Hop Reasoning?</strong><br><br>
Multi-hop reasoning means the answer requires <strong>connecting information from multiple
"hops"</strong> — you cannot answer from a single fact. Each hop retrieves or derives one
piece of information, and only by chaining all the hops together can you reach the final
answer.<br><br>
<strong>Why it matters:</strong> Most real management questions are multi-hop. "Which of our
Tier-1 services that were refactored recently still miss their latency SLO?" requires:
(1) identify services, (2) filter by tier, (3) check refactor dates, (4) check latency.
Without CoT, the model may skip a hop and give a plausible-sounding but wrong answer.
</div>

Here is a slice of a service portfolio at a large retailer. We will ask a question that
requires multiple hops to answer:

<div class="concept-box">
<strong>Service Portfolio Data:</strong>

| Service | Owning Team | Last Major Refactor | p99 Latency (ms) | Tier |
|---------|------------|--------------------|------------------|------|
| Cart Service | Team Atlas | 2024 | 450 | Tier-1 |
| Search Service | Team Nova | 2022 | 380 | Tier-2 |
| Payments Service | Team Meridian | 2025 | 280 | Tier-1 |
| Recommendations | Team Orion | 2023 | 520 | Tier-2 |
| Order History Service | Team Atlas | 2025 | 350 | Tier-1 |
| Pricing Service | Team Meridian | 2021 | 180 | Tier-1 |
| Loyalty Service | Team Nova | 2024 | 390 | Tier-2 |
| Inventory Service | Team Orion | 2024 | 310 | Tier-1 |
</div>

**Question:** "Among the Tier-1 services that were last refactored after 2023, which one has
the **second-lowest** p99 latency?"

This requires chained hops: filter by tier &rarr; filter by refactor date &rarr; collect the
latencies &rarr; sort them &rarr; pick the second-lowest. Skipping or fumbling any hop gives a
plausible-looking wrong answer.

```python
service_data = '''Service portfolio data:
- Cart Service: Team Atlas, last major refactor 2024, p99 latency 450 ms, Tier-1
- Search Service: Team Nova, last major refactor 2022, p99 latency 380 ms, Tier-2
- Payments Service: Team Meridian, last major refactor 2025, p99 latency 280 ms, Tier-1
- Recommendations: Team Orion, last major refactor 2023, p99 latency 520 ms, Tier-2
- Order History Service: Team Atlas, last major refactor 2025, p99 latency 350 ms, Tier-1
- Pricing Service: Team Meridian, last major refactor 2021, p99 latency 180 ms, Tier-1
- Loyalty Service: Team Nova, last major refactor 2024, p99 latency 390 ms, Tier-2
- Inventory Service: Team Orion, last major refactor 2024, p99 latency 310 ms, Tier-1'''

question = '''Among the Tier-1 services that were last refactored after 2023, which one has
the SECOND-lowest p99 latency? Answer with the single service name.'''

# Without CoT
result_direct = ask(service_data + '\n\n' + question)
show('Multi-Hop WITHOUT CoT', result_direct)
```

```python
# With CoT
question_cot = '''Among the Tier-1 services that were last refactored after 2023, which one has
the SECOND-lowest p99 latency? Answer with the single service name.

Let us reason through this step by step:
Step 1: List all Tier-1 services.
Step 2: From those, keep only services last refactored after 2023 (i.e., 2024 or later).
Step 3: Write out the p99 latency of each remaining service.
Step 4: Sort them from lowest to highest p99 latency.
Step 5: State the service with the SECOND-lowest p99 latency.'''

result_cot = ask(service_data + '\n\n' + question_cot)
show('Multi-Hop WITH CoT', result_cot)

print('Expected answer: Inventory Service')
print('  Hop 1 (Tier-1): Cart, Payments, Order History, Pricing, Inventory')
print('  Hop 2 (refactor after 2023): Cart (2024), Payments (2025),')
print('        Order History (2025), Inventory (2024)  —  Pricing (2021) drops out')
print('  Hop 3 (latencies): 450, 280, 350, 310')
print('  Hop 4 (sorted): 280 Payments < 310 Inventory < 350 Order History < 450 Cart')
print('  Hop 5 (second-lowest): Inventory Service (310 ms)')
```

<div class="diagram-box">
<h4>Multi-Hop Reasoning — Chaining Facts to Reach an Answer</h4>

<div style="margin: 20px 0;">
<span class="flow-step" style="background: #d4e8f0; font-weight: bold;">8 Services</span>
<span style="font-size: 20px; vertical-align: middle;">&rarr;</span>
<span class="flow-step">Hop 1: Filter by<br>Tier = Tier-1<br><em>(5 remain)</em></span>
<span style="font-size: 20px; vertical-align: middle;">&rarr;</span>
<span class="flow-step">Hop 2: Filter by<br>refactor &gt; 2023<br><em>(4 remain)</em></span>
<span style="font-size: 20px; vertical-align: middle;">&rarr;</span>
<span class="flow-step">Hop 3+4: Sort by<br>p99 latency<br><em>(280 &lt; 310 &lt; 350 &lt; 450)</em></span>
<span style="font-size: 20px; vertical-align: middle;">&rarr;</span>
<span class="flow-step" style="background: #e8f5e8; border-color: #a8d5a8; font-weight: bold;">Inventory<br>Service (2nd-lowest)</span>
</div>

<p style="font-size: 13px; color: #666; margin-top: 15px;">
Each hop reduces or orders the candidate set. Without explicit steps, the model skipped a
candidate in the filter hop and returned a plausible but incorrect answer (Order History
Service — which is only second-lowest if you forget Inventory Service qualifies).
</p>
</div>

## Part 6: Tree-of-Thought — Multiple Experts Weigh In

<div class="concept-box">
<strong>What is Tree-of-Thought (ToT)?</strong><br><br>
Chain-of-Thought follows a single reasoning path. But for complex decisions, a single
perspective may be biased or incomplete. <strong>Tree-of-Thought</strong> simulates
<strong>multiple independent thinkers</strong>, each analyzing the problem from their own
expertise, and then synthesizes their perspectives into a final recommendation.<br><br>
Think of it like an architecture review board: one reviewer analyzes delivery risk and cost,
another analyzes team capacity, a third analyzes the long-term architecture. No single person
has the full picture, but together they make a better call than any one of them alone.<br><br>
<strong>Key difference from CoT:</strong><br>
<strong>CoT</strong> = one chain of steps (linear, like a single expert)<br>
<strong>ToT</strong> = multiple chains that branch and converge (tree-shaped, like a review panel)
</div>

<div class="diagram-box">
<h4>Tree-of-Thought: Multiple Expert Branches Converging</h4>

<div style="margin: 25px 0;">
<span class="flow-step" style="background: #d4e8f0; font-weight: bold; font-size: 14px;">PROBLEM</span>
<br><br>
<span style="font-size: 22px;">&swarr; &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; &darr; &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; &searr;</span>
<br><br>
<span class="flow-step" style="min-width: 140px;">Expert 1<br><strong>Delivery-Risk Analyst</strong><br><em>cost, timeline,<br>peak-season freeze</em></span>
&nbsp;&nbsp;
<span class="flow-step" style="min-width: 140px;">Expert 2<br><strong>Team-Capacity Analyst</strong><br><em>skills, staffing,<br>attrition risk</em></span>
&nbsp;&nbsp;
<span class="flow-step" style="min-width: 140px;">Expert 3<br><strong>Architecture Strategist</strong><br><em>scalability, coupling,<br>long-term velocity</em></span>
<br><br>
<span style="font-size: 22px;">&searr; &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; &darr; &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; &swarr;</span>
<br><br>
<span class="flow-step" style="background: #e8f5e8; border-color: #a8d5a8; font-weight: bold; font-size: 14px;">SYNTHESIS &amp; RECOMMENDATION</span>
</div>
</div>

```python
# Tree-of-Thought: Strategic decision with multiple experts — refactor vs rewrite
tot_prompt = '''Three independent experts analyze this engineering decision one at a time.
Each expert provides their analysis without seeing the others' work. After all three,
synthesize their perspectives into a final recommendation.

For each expert I want you to come up explaination in few line with the pros and cons of the each step. and suggest the imapct analysis as a 
third point after Pros and cons. and the verdict as the fourth point.

DECISION: Should a large retailer's e-commerce group incrementally refactor its 12-year-old
monolithic order-management service (strangler pattern, domain by domain) or rewrite it from
scratch on the new platform stack?

Context: The service processes every online order. Change-failure rate has doubled in two
years and the service consumes 30% of the group's on-call load. The group has 200 engineers,
a peak-season code freeze starts in 6 months, and two previous "big rewrite" attempts
elsewhere in the company were cancelled midway.

---

EXPERT 1 — Delivery-Risk & Cost Analyst:
Analyze from a delivery perspective: cost and duration of each option, risk of missing the
peak-season freeze, migration/dual-run costs, and the sunk-cost history of cancelled rewrites.

EXPERT 2 — Team-Capacity Analyst:
Analyze from a people perspective: skills available for the new stack vs the legacy stack,
staffing both the old and new systems during transition, on-call burden, morale and
attrition risk on the team that keeps the lights on.

EXPERT 3 — Architecture Strategist:
Analyze from an architecture perspective: how much of the monolith's domain logic is still
correct and worth preserving, coupling hot-spots, scalability limits at peak traffic, and
long-term delivery velocity under each option.

---

SYNTHESIS: After all three experts share their analysis, combine their perspectives into a
clear recommendation with supporting rationale. give the nswers without **, list wise answers are preferred'''

result_tot = ask(tot_prompt, max_tokens=4000, temperature=0.4)
show('Tree-of-Thought: Refactor vs Rewrite', result_tot)
```

## Part 7: ToT for Complex Calculations — The Dock-Door Problem

Tree-of-Thought is not just for strategic decisions. It also works for **complex
calculations** where arithmetic errors are common. The idea: have multiple independent
"solvers" work the same problem, then compare their answers.

<div class="concept-box">
<strong>The Dock-Door Problem:</strong><br><br>
A large retailer's fulfillment center has 120 dock doors. Currently 75% are occupied by
trailers. Tomorrow, 15 trailers will depart, 22 inbound trailers are scheduled, and 8 doors
are closed for maintenance (unavailable). How many doors are free after tomorrow's changes?<br><br>
This problem has 5 quantities that must be tracked correctly. Without structured reasoning,
models frequently forget the maintenance doors or make arithmetic errors.
</div>

<div class="try-it">
<strong>Our approach:</strong> We will have 3 independent "calculators" solve this problem,
then compare their answers. If 2 out of 3 agree, we have confidence in the answer. If all 3
disagree, we investigate. This is the foundation for "LLM-as-a-Judge" (E04).
</div>

```python
dock_problem = '''A fulfillment center has 120 dock doors. Currently 75% are occupied
by trailers. 15 trailers will depart tomorrow. 22 inbound trailers are scheduled tomorrow.
8 doors are closed for maintenance and cannot be used.
How many dock doors are free after tomorrow's changes?'''

# Direct answer (no structured reasoning)
result_direct = ask(dock_problem)
show('Dock-Door Problem — Direct Answer', result_direct)
```

```python
# Three independent solvers with different approaches
solver_system_1 = '''You are Calculator A. Solve the problem by tracking OCCUPIED doors.
Show every step. End with: FINAL ANSWER: [number] doors free.'''

solver_system_2 = '''You are Calculator B. Solve the problem by tracking FREE doors.
Show every step. End with: FINAL ANSWER: [number] doors free.'''

solver_system_3 = '''You are Calculator C. Solve the problem using a table format:
| Category | Count |. Show every step. End with: FINAL ANSWER: [number] doors free.'''

dock_problem = '''A fulfillment center has 120 dock doors. Currently 75% are occupied by
trailers. 15 trailers will depart tomorrow. 22 inbound trailers are scheduled tomorrow.
8 doors are closed for maintenance and cannot be used.
How many dock doors are free after tomorrow's changes?

Think step by step.'''

# Run all three solvers
result_a = ask(dock_problem, system=solver_system_1)
result_b = ask(dock_problem, system=solver_system_2)
result_c = ask(dock_problem, system=solver_system_3)

show('Calculator A (tracking occupied doors)', result_a)
show('Calculator B (tracking free doors)', result_b)
show('Calculator C (table format)', result_c)

print('=' * 60)
print('  EXPECTED ANSWER')
print('=' * 60)
print('Total doors:        120')
print('Occupied (75%):      90')
print('Departures:         -15  =>  occupied becomes 75')
print('Inbound:            +22  =>  occupied becomes 97')
print('Under maintenance:    8  =>  unusable')
print('Free = 120 - 97 - 8 = 15 doors')
print()
print('If 2/3 solvers agree on 15, we have high confidence.')
print('If they disagree, we investigate which step went wrong.')
```

<div class="takeaway">
<strong>Why multiple solvers matter:</strong><br><br>
A single reasoning chain might make an error at any step — and every subsequent step builds
on that error, making it hard to catch. With multiple independent solvers:<br>
<ul>
<li>If 2/3 or 3/3 agree, you have high confidence in the answer</li>
<li>If they disagree, you can compare their steps to find exactly where the error occurred</li>
<li>Different approaches (tracking occupied vs free) serve as cross-checks</li>
</ul>
This "multiple independent solvers + consensus" pattern is the foundation of the
<strong>LLM-as-Judge</strong> technique we will explore in E04.
</div>

1. **LLMs rush to answers** — without reasoning instructions, they pattern-match instead of
   solving, especially when decoy information is present

2. **Chain-of-Thought** forces step-by-step reasoning, where each step becomes context for
   the next. The simplest form is "Let us think step by step" (zero-shot CoT)

3. **Few-shot CoT** provides worked examples when the model needs to follow a specific
   problem-solving strategy it might not discover on its own

4. **CoT transforms diagnostics** — turning a "list of possible causes" into a structured
   "root-cause procedure" with decision points (our checkout latency spike)

5. **Multi-hop reasoning** chains multiple facts together; CoT prevents the model from
   skipping hops and producing plausible-but-wrong answers

6. **Tree-of-Thought** simulates multiple independent experts for strategic decisions
   (refactor vs rewrite) and complex calculations where cross-verification catches errors

7. **Choose the technique to match the problem** — use the decision framework to pick the
   simplest technique that gives reliable results
</div>

<div class="concept-box">
<strong>📖 Glossary:</strong>
<ul>
<li><strong>Chain-of-Thought (CoT)</strong> — prompting the model to write out intermediate
reasoning steps before the final answer</li>
<li><strong>Zero-Shot CoT</strong> — triggering CoT with a bare instruction like "think step
by step", no examples</li>
<li><strong>Few-Shot CoT</strong> — triggering CoT with one or more fully worked examples
that demonstrate the desired reasoning strategy</li>
<li><strong>Tree-of-Thought (ToT)</strong> — running multiple independent reasoning paths
(experts/solvers) on the same problem and synthesizing them</li>
<li><strong>Multi-hop reasoning</strong> — answers that require chaining several facts or
filters; skipping a hop produces plausible-but-wrong answers</li>
<li><strong>Decoy information</strong> — data in a problem that looks relevant but is not;
a common trigger for pattern-matching errors</li>
<li><strong>Consensus / self-consistency</strong> — comparing several independent solutions
and trusting agreement; the seed of LLM-as-a-Judge (E04)</li>
</ul>
</div>

<strong>What does ToT mean for incident triage?</strong><br><br>
For a math problem, ToT meant multiple <em>calculators</em>. For production incidents, ToT
means a <strong>panel of specialist engineers</strong> — each looks at the same ticket through
their own lens, independently, and then an incident commander synthesizes their findings:
<ul>
<li><strong>Expert 1 — Application &amp; Service Specialist:</strong> service code paths,
API contracts, timeouts, retries, cache behavior, feature flags</li>
<li><strong>Expert 2 — Data &amp; Infrastructure Specialist:</strong> databases, queues,
connection pools, network, storage, capacity limits</li>
<li><strong>Expert 3 — Deploy &amp; Platform Specialist:</strong> recent deploys and config
pushes, rollout strategy, platform/runtime versions, autoscaling behavior</li>
</ul>
This is exactly how a real Sev-1 bridge works — no single engineer owns every layer, so the
hardest incidents get a <em>panel</em>.

### Step 1 — Build a reusable "three ways" triage function

This one function is the takeaway of the whole lab. Give it any incident ticket and it runs
the same problem through <strong>no technique</strong>, <strong>CoT</strong>, and
<strong>ToT</strong> so you can compare the quality difference for yourself. Your teams can
reuse it at work — just change the <code>ticket</code> text.

```python
# ── The reusable incident-triage comparison function ──────────────────
# Runs ANY incident ticket through all three approaches so you can compare.

# System prompt for the CoT (single senior engineer, structured methodology)
COT_SYSTEM = '''You are a senior site reliability engineer at a large retailer. For EACH
diagnostic step:
(1) STATE what to check and exactly where (which dashboard, log, trace, or system)
(2) EXPLAIN the system-level reason this check matters for THIS symptom
(3) INTERPRET what each possible result would mean
(4) DECIDE the next step based on the most likely result.
Work from the most likely cause to the least likely — one path at a time,
like a real incident investigation. Read the ticket carefully for clues
(recent changes, what already works, which user segment is affected) and anchor
your investigation on them.'''

# System prompt for the ToT (panel of three specialists, then synthesis)
TOT_SYSTEM = '''You are running a Sev-1 incident escalation panel at a large retailer.
THREE specialist engineers analyze the SAME ticket independently, each from their own
lens, then an incident commander synthesizes.

EXPERT 1 — Application & Service Specialist: service code paths, API contracts, timeouts,
retries, cache behavior, feature flags.
EXPERT 2 — Data & Infrastructure Specialist: databases, queues, connection pools, network,
storage, capacity limits.
EXPERT 3 — Deploy & Platform Specialist: recent deploys/config pushes, rollout strategy,
runtime versions, autoscaling behavior.

Format your answer as:
[EXPERT 1] ...independent analysis with specific checks...
[EXPERT 2] ...independent analysis with specific checks...
[EXPERT 3] ...independent analysis with specific checks...
[INCIDENT COMMANDER SYNTHESIS] The single most likely root cause, the order to check
things, and the fix — reconciling all three experts.'''

def troubleshoot_three_ways(ticket, max_tokens=1800):
    '''Run one incident ticket through: no technique, CoT, and ToT.'''
    print('#' * 68)
    print('  TICKET')
    print('#' * 68)
    print(ticket.strip())

    # 1. No technique — just ask
    r1 = ask(ticket, max_tokens=max_tokens)
    show('APPROACH 1 — NO TECHNIQUE (generic list of causes)', r1)

    # 2. Chain-of-Thought — single structured senior engineer
    r2 = ask(ticket, system=COT_SYSTEM, max_tokens=max_tokens)
    show('APPROACH 2 — CHAIN-OF-THOUGHT (structured diagnostic path)', r2)

    # 3. Tree-of-Thought — panel of three specialists + synthesis
    r3 = ask(ticket, system=TOT_SYSTEM, max_tokens=max_tokens, temperature=0.4)
    show('APPROACH 3 — TREE-OF-THOUGHT (three-specialist panel + synthesis)', r3)

    return r1, r2, r3

print('troubleshoot_three_ways() ready — pass it any incident ticket.')
```

### When should YOUR teams reach for CoT vs ToT on a real ticket?

<table class="compare-table">
<tr><th>Situation</th><th>Use</th><th>Why</th></tr>
<tr>
  <td>Single-domain issue with a clear symptom (one service, one layer)</td>
  <td><strong>Chain-of-Thought</strong></td>
  <td>One structured diagnostic path is faster and cheaper; a panel is overkill</td>
</tr>
<tr>
  <td>Symptom could live in several layers (edge + service + data)</td>
  <td><strong>Tree-of-Thought</strong></td>
  <td>Independent specialists each rule their layer in/out; catches cross-layer causes</td>
</tr>
<tr>
  <td>A recent change is named in the ticket (deploy, config push, new writer)</td>
  <td><strong>CoT</strong>, anchored on the change</td>
  <td>The trigger already narrows the search — you need depth, not breadth</td>
</tr>
<tr>
  <td>Two engineers already disagree on the root cause</td>
  <td><strong>ToT</strong></td>
  <td>The panel surfaces both hypotheses side-by-side, then the synthesis reconciles them</td>
</tr>
<tr>
  <td>You need a defensible answer for a Sev-1 bridge / RCA document</td>
  <td><strong>ToT</strong></td>
  <td>The multi-expert structure shows you considered every angle, not just the obvious one</td>
</tr>
</table>

<div class="takeaway">
<strong>Rule of thumb:</strong> CoT is your <em>default</em> — most tickets are single-path.
Escalate to ToT when the problem crosses layers, when two experts disagree, or when the
answer has to survive scrutiny in a root-cause review. The cost is real: ToT produces 3
analyses + a synthesis, so it uses more tokens than a single CoT pass. Match the technique to
the stakes.
</div>
